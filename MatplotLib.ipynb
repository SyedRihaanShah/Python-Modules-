```python
#This is my matplotlib practice/learning 
import matplotlib.pyplot as plt 
import numpy as np
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 20, 30 ,40])
y2 = np.array([23, 54, 32, 10])
```

```python
# Plot customization 
plt.plot(x,y1, marker= '.',  markersize = 10,
          markerfacecolor= "cyan",
          mec= 'black',
          linestyle="dashed",
          linewidth = 2,
          color= 'red')
plt.plot(x, y2)
plt.show()
```

```python
#one more way to do 
styles = dict( marker= '.',  markersize = 10,
          markerfacecolor= "cyan",
          mec= 'black',
          linestyle="dashed",
          linewidth = 2,
          color= 'red' )
plt.plot(x, y1, **styles) #**styles means unpacking the dict
plt.plot(x,y2, **styles)
plt.show()
```

LABELS

```python

plt.plot(x,y1)
plt.plot(x,y2)

plt.title("Graph", fontsize = 30,
          family= 'Arial',
          fontweight = 'bold',
          color = 'purple')

plt.xlabel('Year', fontsize = 20 ,
           family = 'Arial',
           fontweight = 'bold',
           color= 'yellow')

plt.ylabel('Studnets', fontsize = 20 ,
           family = 'Arial',
           fontweight = 'bold',
           color= 'red')

plt.tick_params(axis="both", color='pink')

plt.xticks(x)
plt.show()
```

GRID LINES 

```python
plt.plot(x,y1)
# plt.grid()
plt.grid(axis="x")
plt.grid(axis='y', linewidth= 2,
         color = 'black',
         linestyle = 'dashed')

plt.show()
```

BAR CHARTS

```python
categories = np.array(['grains', 'bread', 'fruit', 'dairy', 'sweets'])
values = np.array([4 , 3 , 2, 5, 1])
plt.bar(categories, values, color= 'cyan') # vertical bar chart
#plt.barh(categories, values) # horizontal bar chart
plt.title("Daily consumption")
plt.xlabel('Food')
plt.ylabel('Consumption')
```

PIE CHARTS

```python
from pickle import TRUE


name = np.array(["freshman", "juniours", "seniors"])
values = np.array([300, 250, 225])
colors = np.array(["red", "blue", "green"])

plt.pie(values, labels=name, 
        autopct="%1.1f",
        colors= colors,
        explode=[0,0,0.1],
        shadow=TRUE,)

plt.title("Collage stats")

plt.show()
```

HISTOGRAMS

```python
scores = np.random.normal(loc=80, scale=10, size = 100)
scores = np.clip(scores, 0, 100)
plt.hist(scores, bins= 10,
         color = "lightgreen",
         edgecolor = 'black')
plt.title('histogram')
plt.xlabel('score')
plt.ylabel('no of students')
plt.show()
```

SUBPLOTS

- Figure = The entire canvas
- Ax = a single plot (subplot)

```python
x = np.array([1,2,3,4,5])
figure , axis = plt.subplots(2,2)
axis[0,0].plot(x,x*2)
axis[0,0].set_title('x, x *2')
axis[0,1].plot(x+6, x*2, color = 'green')
axis[1,0].plot(x * x * x, x + 4, color = 'black')
axis[1,1].plot(x *x , x +1, color= 'red')
plt.show()
```
