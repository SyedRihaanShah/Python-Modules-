### HEAT MAP

```python
columns = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']

df[columns].corr()
```

```python
sns.heatmap(df[columns].corr(), annot= True, cmap='Blues', vmin= -0.2, linewidths=1.6, linecolor='Black')
```

### RUG PLOT 

```python
sns.rugplot(df, x = 'body_mass_g',hue='species', palette= 'Set2', height= 0.4)
```

### PAIR PLOT

```python
sns.pairplot(data = df[columns], palette= 'Set2', diag_kind='hist',)
```

### Pair Grid

```python
graph = sns.PairGrid(df, hue = 'sex', palette= 'Set2')
graph.map_upper(sns.scatterplot)
graph.map_lower(sns.kdeplot)
graph.map_diag(sns.histplot)
graph.add_legend()
```
